# Heron's Formula
For a triangle of sidelengths $a, b, c$, we define the semi-perimeter of the triangle, $s = \frac{1}{2}(a+b+c)$, and the area of the triangle is $$A = \sqrt{s(s-a)(s-b)(s-c)}$$

So if we have sides of length $n, n, n \pm 1$, the area would be $$A = \frac{n\pm 1}{2}\sqrt{(3n \pm 1)(n \mp 1)}$$
