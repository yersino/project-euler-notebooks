# Problem 088
https://projecteuler.net/problem=88

We try another hard problem. This is our first Level $7$ problem in Project Euler and asks us to find the sum of all minimal product-sum numbers for $2 \leq k \leq 12000$.

As a recap, a natural $N$ is a product-sum number if it can be given as both the product and sum of a given set of at least two natural numbers, whose size is $k$, i.e. $$N = a_1 \times a_2 \times a_k = a_1 + a_2 + \dots + a_k$$.

Project Euler is nice enough to give us the minimal product-sum number for sets of size $k = 2, 3, 4, 5, 6$. I don't want to write it all out but do notice that for $k \geq 3$.
