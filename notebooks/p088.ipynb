# Problem 088
https://projecteuler.net/problem=88

We try another hard problem. This is our first Level $7$ problem in Project Euler and asks us to find the sum of all minimal product-sum numbers for $2 \leq k \leq 12000$.

## Problem Statement and First Thoughts

As a recap, a natural $N$ is a product-sum number if it can be given as both the product and sum of a given set of at least two natural numbers, whose size is $k$, i.e. $$N = a_1 \times a_2 \times a_k = a_1 + a_2 + \dots + a_k$$
call this set the representitive $k-$set. Order elements so that it is stirctly non-increasing. Call the minimal product-sum number $N$ the $k-$target.

Project Euler is nice enough to give us the minimal product-sum number for sets of size $k = 2, 3, 4, 5, 6$. I don't want to write it all out but do notice that for $k \geq 3$, the representitive set has multiple $1$'s. If we're trying to find the smallest sum possible, this is sensible. It also helps reduce the problem's complexity: rather than finding $k$ numbers that sum and multiply to $N$, we can instead try to find $k-w$ numbers that sum to $N-w$ and multiply to $N$ for $w$ $1$'s. Say we also have $v$ non-unit factors in our representative set too. 

Our work flow is to find the sum of the set all minimal product-sum numbers for $2 \leq k \leq 12000$. 

So we expect roughly the following functions:
- For each $k$, we want to iterate up from some lower bound, trying to keep as many entries 1's as possible (or start off with N-k-1 and try to find factors that multiply to that fior increasing increment)
- For each $k-$target, we want to add it to a certain set of $k-$ targets. 
- Sum up every element in that set. 

But eagle-eyed among you may note that this would be incredibly expensive to compute. Doing this for $5-$ digit $k$'s would take insane amounts of time. So what can we do instead? 

## Generating Factorisations
 
It might seem trivial to remark that every representative set would be associated with a unique target. Additionally, we can try to focus on only the non-unit elements given that $$N = \displaystyle \prod^{k}_{i=1}a_i = \prod^{v}_{i=1}a_i = \sum_{i=1}^{v}a_i + w$$ hence $$k = v + w = v + N - \sum_{i=1}^{v}a_i$$
