## First Thoughts
We may try doing this by hand and quickly realise that it's just a constraint satisfaction problem. 

We must have one copy of a $0$ in one set and $1$ in another. We must have one copy of a $0$ and a $4$ in the other. We can keep proceeding in this way and see that the required multiplets are in fact:


```python
requiredPairs = [
    (0,1),
    (0,4),
    (0,9),   # 6 works too after identifying 6, 9
    (1,6),
    (2,5),
    (3,6),
    (4,9),
    (6,4),
    (8,1),
]
```

So we have 9 conditions and need to use these, either pairing or splitting them up to generate unique possibilities.

This is very difficult!

Can we do something else?
 
It would be nice if we could instead iterate over possible combinations as a brute-force rather than constructing them. So let's give it a shot. 

As a first check, each cube has 6 distinct digits chosen from $\{0, 1, 2, 3, 4, 5, 6, 7, 8\}$, so we anticipate that there are $9$ choose $6 = 84$ possible combinations for each cube. We may double this if we identify $6, 9$ independently, giving $168$ combinations, and hence $\frac{168 \times 169}{2} = 14196$ cube pairs. 

Luckily, this is not so big! Iterating over possible combinations is feasible!  

So let's draft a helper function to check if a combination is achievable. 

```python
def possibleCombination(num1, num2, cube1, cube2):
    return ((num1 in cube1 and num2 in cube2) or (num2 in cube1 and num1 in cube2))
```

so our final code is just:

```python
from itertools import combinations

digits = range(10)

# we get all possible cube combinations
cubes = []
for c in combinations(digits, 6):
    s = set(c)
    if 6 in s or 9 in s:
        s |= {6,9}
    cubes.append(s)

ans = 0

#and then we check them !
for i in range(len(cubes)):
    for j in range(i, len(cubes)):      # unordered pairs
        if all(possibleCombination(x, y, cubes[i], cubes[j])
               for x, y in requiredPairs):
            ans += 1

print(ans)
```
